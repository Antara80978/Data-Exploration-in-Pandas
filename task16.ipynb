```python
# This confirms if your data classes are balanced
print(df_iris['species'].value_counts())
```

```python
 #This reveals the distinct size differences between species
print(df_iris.groupby('species').mean())
```

```python
# Shows if longer petals also mean wider petals
print(df_iris.corr(numeric_only=True))
```

**Titanic Dataset**

```python
# Load the dataset
df_titanic = pd.read_csv("Titanic-Dataset.csv")
```

```python
print(df_titanic.shape)
print(df_titanic.head())
```

```python
#see 'age', 'deck', and 'embarked' have missing entries
print(df_titanic.isnull().sum())
```

```python
#Prints the proportion of people who survived (1) vs died (0)
print(df_titanic['Survived'].value_counts(normalize=True))
```

```python
# Check survival rates by Gender
print(df_titanic.groupby('Sex')['Survived'].mean())
```

```python
# Check survival rates by Ticket Class (1st, 2nd, 3rd)
print(df_titanic.groupby('Pclass', observed=False)['Survived'].mean())
```

```python
# Helps see if families or groups are sharing identical ticket data profiles
print(f"Duplicate rows: {df_titanic.duplicated().sum()}")
```
